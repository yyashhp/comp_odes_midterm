## The exponential of a matrix
Let $A$ be an $n\times n$ matrix.  Recall that
$$
    X(t) = e^{At} = I + At + {A^2t^2\over 2} + {A^3t^3\over 3!} +\ldots = \sum_{k=0}^{\infty} {A^kt^k\over k!}
$$
is a fundamental matrix solution of
$$
    \boldsymbol y' = A\boldsymbol y
$$
and
$$
    \left(X(t)\right)^{-1} = X(-t) = e^{-At}.
$$

## Canonical Forms
Suppose $A$ is similar to $J$.  That is, suppose there is an invertible matrix $P$ such that 
$$
    AP = PJ\qquad\hbox{or}\qquad A = PJP^{-1}
$$
then
$$
    e^{At} = P e^{Jt} P^{-1}.
$$

## Complex Eigenvalues
Suppose
$$
    J = \begin{bmatrix} \alpha & \beta\cr -\beta & \alpha \cr\end{bmatrix}
    = \alpha \begin{bmatrix} 1 & 0 \\ 0 & 1\end{bmatrix} 
     + \beta \begin{bmatrix} 0 & 1\cr -1 & 0 \cr\end{bmatrix} = \alpha I + \beta B.
$$
Since the identity matrix commutes with any $2\times2$ matrix,
$$
    e^{Jt} = e^{\alpha I t} e^{B t}.
$$
Furthermore
$$
    e^{Jt} = e^{\alpha I t}e^{Bt} = e^{\alpha t}\begin{bmatrix}
    \cos \beta t & \sin\beta t\cr
    -\sin\beta t & \cos\beta t\cr
    \end{bmatrix}
$$

## Block Diagonal Matrices

If $J$ is a block diagonal matrix, say
$$
    J = \begin{bmatrix}
        J_1 & 0 & 0 & \cdots & 0 \\
        0 & J_2 & 0 & \cdots & 0 \\
        0 & 0 & J_3 & \cdots & 0 \\
        \vdots & \vdots & \vdots & \ddots & \vdots \\
        0 & 0 & 0 &\cdots &J_m\\
    \end{bmatrix}
$$
where each $J_i$ is a square matrix, then

$$
    J^kt^k = \begin{bmatrix}
        J_1^k t^k & 0 & 0 & \cdots & 0 \\
        0 & J_2^k t^k & 0 & \cdots & 0 \\
        0 & 0 & J_3^k t^k & \cdots & 0 \\
        \vdots & \vdots & \vdots & \ddots & \vdots \\
        0 & 0 & 0 &\cdots &J_m^k t^k\\
    \end{bmatrix}
$$

and
$$
    e^{Jt} = \sum_{k=0}^{\infty} {J^k t^k\over k!}
    =\begin{bmatrix} 
        e^{J_1 t} & 0 & 0 & \cdots & 0 \\
        0 & e^{J_2 t} & 0 & \cdots & 0 \\
        0 & 0 & e^{J_3 t} & \cdots & 0 \\
        \vdots & \vdots & \vdots & \ddots & \vdots \\
        0 & 0 & 0 &\cdots &e^{J_m t}\\
    \end{bmatrix}
$$

## Differential Equation:
$$
    y^{(8)}+8y^{(7)}+22y^{(6)}+22y^{(5)}+7y^{(4)}+34y^{(3)}+90 y^{(2)}+80 y'+24 y= 0
$$
Equivalent first order system
\begin{align*}
\dot y = \dot y_1 &= y_2\cr
\dot y_2 &= y_3\cr
&\vdots\cr
\dot y_7 &=y_8\cr
\dot y_8 &= - 8y_8-22y_7-22y_6-7y_5-34y_4-90y_3-80y_2-24y_1\cr
\end{align*}

```python
A = np.eye(8,8,1,dtype=int)
A[-1] = [-24,-80,-90,-34,-7,-22,-22,-8]
A
```

```python
λ,V = LA.eig(A)
λ
```

```python
LA.norm(LA.inv(V)@V-np.eye(8),1)
```

```python
LA.norm(A@V - V@np.diag(λ),2) 
```

### Eigenvalue -1
This eigenvalue has algebraic multiplicity 3, but what is the geometric multiplicity?  This must be the dimension of the nullspace of 
$A+I.$

```python
v1 = V.T[5]
v1/v1[-1]
```

```python
w1 = 2*np.array([-1,1,-1,1,-1,1,-1,1])
A@w1 + w1
```

```python
I8 = np.eye(8,dtype=int)
LA.matrix_rank(A+I8)
```

```python
LA.matrix_rank(LA.matrix_power(A+I8,2))
```

```python
v2 = LA.solve((A+I8),w1)
v2
```

```python
w2 = np.array([-3,1,1,-3,5,-7,9,-11])
(A+I8)@w2 - w1
```

```python
LA.matrix_rank(LA.matrix_power(A+I8,3))
```

```python
v3 = LA.solve(A+I8,w2)
v3
```

```python
w3 = np.array([-3,0,1,0,-3,8,-15,24])
(A+I8)@w3 - w2
```

```python
J1 = np.array([[-1,1,0],[0,-1,1],[0,0,-1]])
J1
```

```python
w1w2w3 = np.vstack((w1,w2,w3)).T
w1w2w3
```

```python
A@w1w2w3 - w1w2w3@J1
```
