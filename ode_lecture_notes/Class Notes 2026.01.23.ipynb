## The exponential of a matrix
Let $A$ be an $n\times n$ matrix.  Then
$$
    e^{At} = I + At + {A^2t^2\over 2} + {A^3t^3\over 3!} +\ldots = \sum_{k=0}^{\infty} {A^kt^k\over k!}.
$$
is a fundamental matrix solution of
$$
    \boldsymbol y' = A\boldsymbol y
$$
with
$$
    e^{A0} = I\qquad\hbox{and}\qquad\left(e^{At}\right)^{-1} = e^{-At}.
$$

## Canonical Forms
Suppose $A$ is similar to $J$.  That is, there is an invertible matrix $P$ such that 
$$
    A = PJP^{-1}
$$
or
$$
    AP = PJ
$$

Then
$$
    e^{At} = I + At+ {A^2t^2\over 2} + {A^3t^3\over 3!} +\cdots
    =  PP^{-1}+ PJtP^{-1} + {PJ^2t^2P^{-1}\over 2} +\cdots = P e^Jt P^{-1} $$
    

## Complex Eigenvalues
Suppose
$$
    J = \begin{bmatrix} \alpha & \beta\cr -\beta & \alpha \cr\end{bmatrix}
    = \alpha \begin{bmatrix} 1 & 0 \\ 0 & 1\end{bmatrix} 
     + \beta \begin{bmatrix} 0 & 1\cr -1 & 0 \cr\end{bmatrix} = \alpha I + \beta B.
$$
Since the identity matrix commutes with any $2\times2$ matrix,
$$
    e^{Jt} = e^{\alpha I t} e^{B t}.
$$
Furthermore
$$
    e^{Jt} = e^{\alpha I t}e^{Bt} = e^{\alpha t}\begin{bmatrix}
    \cos \beta t & \sin\beta t\cr
    -\sin\beta t & \cos\beta t\cr
    \end{bmatrix}
$$

## Simpler fundamental matrix.
If $Q$ is an invertible matrix, then the columns of
$$
    X(t) = e^{At} Q
$$
are
1. linearly independent for all $t$, and
2. are all linear combinations of the columns of $e^{At}$.
It follows that $X(t)$ is a fundamental matrix solution of 
$$
    y' = Ay
$$

In particular,
$$
    X(t) = e^{At}P = Pe^{Jt}
$$
is also a fundamental matrix solution.

## Example
No basis of eigenvectors

```python
A = np.array([[1,1],[0,1]])
A
```

Notice that one is the only eigenvalue and
$$
A  - I = \begin{bmatrix} 0 & 1\cr 0 & 0 \end{bmatrix}
$$
has a one dimensional null space spanned by
$$
    \begin{bmatrix} 1 \\ 0 \end{bmatrix}.
$$

This is a standard example of a matrix without a basis of eigenvectors.

Notice
$$
    A = I + \begin{bmatrix} 0 & 1\cr 0 & 0 \end{bmatrix} = I + N
$$
with
$$
    N^2 = 0.
$$

It follows that
$$
    e^{Nt} = I + Nt + 0
$$
and 
$$
    e^{At} = e^{t}e^{Nt}.
$$

```python
A = Matrix([[1,1],[0,1]])
expAt = exp(t)*Matrix([[1,t],[0,1]])
expAt
```

```python
diff(expAt,t)-A*expAt
```
