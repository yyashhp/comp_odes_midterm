## The exponential of a matrix
Let $A$ be an $n\times n$ matrix.  Recall that
$$
    X(t) = e^{At} = I + At + {A^2t^2\over 2} + {A^3t^3\over 3!} +\ldots = \sum_{k=0}^{\infty} {A^kt^k\over k!}
$$
is a fundamental matrix solution of
$$
    \boldsymbol y' = A\boldsymbol y
$$
and
$$
    \left(X(t)\right)^{-1} = X(-t) = e^{-At}.
$$

## Canonical Forms
Suppose $A$ is similar to $J$.  That is, suppose there is an invertible matrix $P$ such that 
$$
    AP = PJ\qquad\hbox{or}\qquad A = PJP^{-1}
$$
then
$$
    e^{At} = P e^{Jt} P^{-1}.
$$

## Complex Eigenvalues
Suppose
$$
    J = \begin{bmatrix} \alpha & \beta\cr -\beta & \alpha \cr\end{bmatrix}
    = \alpha \begin{bmatrix} 1 & 0 \\ 0 & 1\end{bmatrix} 
     + \beta \begin{bmatrix} 0 & 1\cr -1 & 0 \cr\end{bmatrix} = \alpha I + \beta B.
$$
Since the identity matrix commutes with any $2\times2$ matrix,
$$
    e^{Jt} = e^{\alpha I t} e^{B t}.
$$
Furthermore
$$
    e^{Jt} = e^{\alpha I t}e^{Bt} = e^{\alpha t}\begin{bmatrix}
    \cos \beta t & \sin\beta t\cr
    -\sin\beta t & \cos\beta t\cr
    \end{bmatrix}
$$

## Block Diagonal Matrices

If $J$ is a block diagonal matrix, say
$$
    J = \begin{bmatrix}
        J_1 & 0 & 0 & \cdots & 0 \\
        0 & J_2 & 0 & \cdots & 0 \\
        0 & 0 & J_3 & \cdots & 0 \\
        \vdots & \vdots & \vdots & \ddots & \vdots \\
        0 & 0 & 0 &\cdots &J_m\\
    \end{bmatrix}
$$
where each $J_i$ is a square matrix, then

$$
    J^kt^k = \begin{bmatrix}
        J_1^k t^k & 0 & 0 & \cdots & 0 \\
        0 & J_2^k t^k & 0 & \cdots & 0 \\
        0 & 0 & J_3^k t^k & \cdots & 0 \\
        \vdots & \vdots & \vdots & \ddots & \vdots \\
        0 & 0 & 0 &\cdots &J_m^k t^k\\
    \end{bmatrix}
$$

and
$$
    e^{Jt} = \sum_{k=0}^{\infty} {J^k t^k\over k!}
    =\begin{bmatrix} 
        e^{J_1 t} & 0 & 0 & \cdots & 0 \\
        0 & e^{J_2 t} & 0 & \cdots & 0 \\
        0 & 0 & e^{J_3 t} & \cdots & 0 \\
        \vdots & \vdots & \vdots & \ddots & \vdots \\
        0 & 0 & 0 &\cdots &e^{J_m t}\\
    \end{bmatrix}
$$

## Technique used in an introductory differential equations course.

We typically give students the following "cookbook" approach to solving this problem.

### Starting with the differential equation
Differential equation
$$
    y^{(8)}+8y^{(7)}+22y^{(6)}+22y^{(5)}+7y^{(4)}+34y^{(3)}+90\ddot y+80\dot y+24 y = 0.
$$
form the **characteristic polynomial of the differential equation**
$$
    p(r) = r^{8} + 8 r^{7} + 22 r^{6} + 22 r^{5} + 7 r^{4} + 34 r^{3} + 90 r^{2} + 80 r + 24
$$
Notice that this is the same as the characteristic polynomial of the matrix $A$.

```python
r = Symbol('r')
p = Matrix(A).charpoly(r).as_expr()
p
```

```python
p.factor()
```

```python
y1,y2,y3 = exp(-t),t*exp(-t),t**2*exp(-t)
y4,y5 = exp(-2*t),t*exp(-2*t)
y6 = exp(-3*t)
y7,y8 = exp(t)*cos(t),exp(t)*sin(t)
```

```python
row = [y1,y2,y3,y4,y5,y6,y7,y8]
rows = [row]
for k in range(1,8):
    row = [diff(yi,t) for yi in rows[-1]]
    rows.append(row)
X = Matrix(rows)
X
```

```python
X.det().simplify()
```

```python
diff(X,t)-A@X
```

## Hartman-Grobman Theorem
$\newcommand{\xstar}{x^{\star}}$
$\newcommand{\R}{{\Bbb R}}$
https://en.wikipedia.org/wiki/Hartman–Grobman_theorem

### Hypotheses
1. Let $U$ be an **open neighborhood** of $x^{\star}$ in ${\Bbb R}^n$.  That is $U$ is open (all points of $U$ are interior points) and contains $x^{\star}$.

2. Let $f:U\to{\Bbb R}^n$ be $C^{1}$ on $U$ (continuous partial derivatives of the components w.r.t. the usual coordinates).

3. Let $J$ be the Jacobian matrix
$$
    J = [J_{ij}] = [\partial f_i/\partial x_j].
$$

4. Let $\phi(t,y_0)=\phi^t(y_0)=\phi_{y_0}(t)$ be the **flow** for 
\begin{equation}
    \dot y = f(y),\qquad y(0)=y_0. \tag{IVP}
\end{equation}
That is, for each initial condition $y_0\in U$, $\phi_{y_0}(t)$ satisfies (IVP) on some open interval about $t=0$. 

5. Suppose $x^{\star}$ is an **equilibrium point** (or **fixed point**) for $f$.  That is, $f(x^{\star})=0$ hence
$$
    \phi(t,x^{\star}) \equiv x^{\star}.
$$

6. Suppose $A=J(x^{\star})$ is **hyperbolic**.  That is, all eigenvalues of $A$ have nonzero real part.

### Conclusion
#### Then
1. there is a **neighborhood** $N$ of $x^{\star}$ (a set with $x^{\star}$ in its interior), 

2. a neighborhood $V$ of $0$ in ${\Bbb R}^n$, and

3. a **homeomorphism** $h:N\to V$, that is $h$ is continuous, invertible, and its inverse is continuous

#### such that
* under the change of coordinates $x=h(y)$ the flow $\phi$ is topologically conjugate to the flow for the linear system
$$
    \dot x = Ax\qquad x(t_0) = x_0
$$
which as we now know is $e^{At}x_0$.  That is, $h(\phi(t,y_0))=e^{At}x_0$ with $x_0=h(y_0)$.
